```python
import json
import csv
import os

src = '../Datasets_ClimateVisions-main'
file_name = ['climate_ct_new.json', 'climate_tv_new.json']
# Input JSON file
for f in file_name:
    json_file = f"{src}/{f}"

    # Output CSV file
    csv_file = f.replace('.json', '.csv')

    # Load JSON data
    with open(json_file, "r", encoding="utf-8") as f:
        data = json.load(f)

    # If JSON is a single object, convert it to a list
    if isinstance(data, dict):
        data = [data]

    # Get CSV headers from JSON keys
    headers = data[0].keys()

    # Write CSV
    with open(csv_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        writer.writerows(data)

print(f"CSV file saved as: {os.path.abspath(csv_file)}")
```

```python
import pandas as pd
import os

src = '../Filter_data'

files = [
    'climate_ct_new.csv',
    'climate_tv_new.csv'
]

total_deleted = 0
total_before = 0
total_after = 0

for file in files:
    path = os.path.join(src, file)

    df = pd.read_csv(path)

    before = len(df)

    # normalize missing values
    df["label_consequences"] = df["label_consequences"].astype(str).str.strip().str.lower()

    # remove all invalid cases
    df_clean = df[
        (df["label_consequences"] != "n/a") &
        (df["label_consequences"] != "na") &
        (df["label_consequences"] != "") &
        (df["label_consequences"] != "nan") &
        (df["label_consequences"] != "no consequences") 
    ]

    after = len(df_clean)

    deleted = before - after

    total_before += before
    total_after += after
    total_deleted += deleted

    output_path = os.path.join(src, file.replace(".csv", "_clean.csv"))
    df_clean.to_csv(output_path, index=False)

    print(f"\n{file}")
    print(f"Before: {before}")
    print(f"After: {after}")
    print(f"Deleted: {deleted}")


```
