```python

import pandas as pd
import os

src = '../Filter_data'

files = [
    'climate_ct_new_clean.csv',
    'climate_tv_new_clean.csv'
]

all_classes = set()
all_counts = []

for file in files:
    path = os.path.join(src, file)
    df = pd.read_csv(path)

    # Count the frequency of each label
    counts = df['label_consequences'].value_counts()

    print(f"\nLabel frequencies in {file}:")
    print(counts.to_string())

    # Collect all unique labels
    all_classes.update(counts.index)

    # Store frequencies for combining datasets
    all_counts.append(counts)

# Combine frequencies across both datasets
total_counts = (
    pd.concat(all_counts, axis=1)
    .fillna(0)
    .sum(axis=1)
    .astype(int)
)

print("\nTotal label frequencies across both datasets:")
print(total_counts.sort_index().to_string())

print(f"\nTotal unique classes: {len(all_classes)}")


# Check whether "..." exists as an actual label
print(df[df['label_consequences'].astype(str).str.contains(r'\.\.\.', regex=True)]['label_consequences'].unique())
```
