```python
import pandas as pd
import os

src = '../Filter_data'  

files = [
    'climate_ct_new_clean.csv',
    'climate_tv_new_clean.csv'
]

all_classes = set()

for file in files:
    path = os.path.join(src, file)

    df = pd.read_csv(path)

    # Get unique values from consequence column
    classes = df['label_consequences'].dropna().unique()

    print(f"\nClasses in {file}:")
    for c in sorted(classes):
        print("-", c)

    # Add to global set
    all_classes.update(classes)

# Print all unique classes across both datasets
print("\n\nAll unique classes across both datasets:")
for c in sorted(all_classes):
    print("-", c)

print(f"\nTotal unique classes: {len(all_classes)}")
```
